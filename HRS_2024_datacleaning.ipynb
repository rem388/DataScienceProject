### **Cleaning Health and Retirement Study (HRS) 2024**

```python
hrs_leisure_raw = pd.read_csv("data/hrs_data/h24lb_r.csv", dtype={"HHID": str, "PN": str})
hrs_cognition_raw = pd.read_csv("data/hrs_data/h24d_r.csv", dtype={"HHID": str, "PN": str})
hrs_health_raw = pd.read_csv("data/hrs_data/h24c_r.csv",dtype={"HHID": str, "PN": str})
hrs_demographics_raw = pd.read_csv("data/hrs_data/h24a_r.csv",dtype={"HHID": str, "PN": str})

display(duckdb.sql("SELECT * FROM hrs_leisure_raw LIMIT 5").df())
display(duckdb.sql("SELECT * FROM hrs_cognition_raw LIMIT 5").df())
display(duckdb.sql("SELECT * FROM hrs_health_raw LIMIT 5").df())
display(duckdb.sql("SELECT * FROM hrs_demographics_raw LIMIT 5").df())
```

We used HRS's 2024 [User Guide](data/hrs_data/h24dd.pdf) to determine which are the relevant variable to our research question in the dataframes. The HRS Core is divided into separate files based on questionnaire section. 
files: the Leave-Behind Questionnaire, Cognition, Physical Health, and Demographics.

### Relevant Variables
**Participant identifiers**
- `HHID`: household identification number
- `PN`: respondent person number within the household
When importing data we made both text strings because HRS specifically warns that identifiers such as HHID and PN may contain leading zeros and that users should preserve them when importing the data. 

**Leave-Behind Questionnaire (`h24lb_r.csv`)**
- `TLB001E`: frequency of attending an educational or training course
- `TLB001I`: frequency of reading books, magazines, or newspapers
- `TLB001J`: frequency of watching television
- `TLB001N`: frequency of using a computer for email, Internet, or other tasks
- `TLB001S`: frequency of playing sports or exercising
- `TLB001T`: frequency of walking for 20 minutes or more

The Leave-Behind Questionnaire is a self-administered survey that collects information about respondents' psychosocial characteristics and lifestyle activities. Although the questionnaire was introduced in 2004, in 2006 the design randomly selected approximately half of HRS households to respond, while the other half were selected in 2008. This alternating design continues across survey waves according to the User Guide.

Because the questionnaire is administered to a selected subset of respondents rather than the entire HRS sample, we expect a substantial number of missing values in these variables. However, not all selected respondents necessarily complete the questionnaire, and additional eligibility requirements may affect participation.

**Cognition (`h24d_r.csv`)**
- `TD174`: number of words correctly recalled during the immediate word-recall task
- `TD184`: number of words correctly recalled during the delayed word-recall task

**Physical Health (`h24c_r.csv`)**
- `TC083`: frequency of having trouble falling asleep
- `TC084`: frequency of waking during the night
- `TC085`: frequency of waking too early and being unable to fall asleep again
- `TC086`: frequency of feeling rested when waking in the morning
- `TC303`: minutes reported daily sitting time
- `TC304`: hours reported daily sitting time

**Demographics (`h24a_r.csv`)**
- `TA019`: respondent's current age

We will now create four smaller dataframes containing only the variables relevant to our analysis and rename the variables to make them easier to interpret.


```python
hrs_leisure = duckdb.sql("SELECT HHID AS hhid, PN AS pn, " \
"TLB001E AS education_training, " \
"TLB001I AS reading, TLB001J AS watching_tv, " \
"TLB001N AS computer_internet, TLB001S AS exercising, " \
"TLB001T AS walking FROM hrs_leisure_raw").df()

hrs_cognition = duckdb.sql("SELECT HHID AS hhid, PN AS pn, " \
"TD174 AS immediate_recall, " \
"TD184 AS delayed_recall FROM hrs_cognition_raw").df()

hrs_health = duckdb.sql("SELECT HHID AS hhid, PN AS pn, " \
"TC083 AS trouble_falling_asleep, " \
"TC084 AS waking_at_night, " \
"TC085 AS waking_too_early, " \
"TC086 AS feeling_rested, " \
"TC303 AS sitting_minutes, " \
"TC304 AS sitting_hours FROM hrs_health_raw").df()

hrs_demographics = duckdb.sql("SELECT HHID AS hhid, PN AS pn, " \
"TA019 AS age FROM hrs_demographics_raw").df()

display(hrs_leisure.head())
display(hrs_cognition.head())
display(hrs_health.head())
display(hrs_demographics.head())
```

### Combining the HRS Dataframes

Although these variables are stored in separate CSV files, the files are all components of the same 2024 Health and Retirement Study data collection. HRS
separates its survey into files based on questionnaire section, such as leisure activities, cognition, physical health, and demographics.

Each respondent-level file contains `HHID` and `PN`. `HHID` identifies the respondent's household, while `PN` identifies the individual respondent within
that household. Together, these variables uniquely identify a respondent.

```python
hrs = duckdb.sql("SELECT l.*, c.* EXCLUDE (hhid, pn), h.* EXCLUDE (hhid, pn), d.* EXCLUDE (hhid, pn) FROM hrs_leisure AS l " \
"INNER JOIN hrs_cognition AS c USING (hhid, pn) " \
"INNER JOIN hrs_health AS h USING (hhid, pn) " \
"INNER JOIN hrs_demographics AS d USING (hhid, pn)").df()

display(hrs.head())
```

### Overall Summary of the HRS Data

Before analyzing relationships between leisure activities and health or cognitive outcomes, we first summarize the combined HRS dataset. We calculate the number of participants and examine the frequency and percentage of responses for our main categorical variables, such as reading, television viewing, exercise, and sleep difficulties.

These statistics help us understand the composition of our sample and the distribution of each variable. They also allow us to identify uncommon, missing, or unusual response values that may need to be addressed before further analysis.

```python
num_participants = duckdb.sql("SELECT COUNT(DISTINCT (hhid, pn)) FROM hrs").fetchone()[0]
print("Number of participants:", num_participants)

reading_frequency = duckdb.sql("SELECT reading, COUNT(*) " \
"AS frequency, ROUND(COUNT(*) * 100.0 / SUM(COUNT(*)) OVER (), 2) " \
"AS percentage FROM hrs WHERE reading IS NOT NULL " \
"GROUP BY reading ORDER BY reading").df()

tv_frequency = duckdb.sql("SELECT watching_tv, COUNT(*) " \
"AS frequency, ROUND(COUNT(*) * 100.0 / SUM(COUNT(*)) OVER (), 2) " \
"AS percentage FROM hrs WHERE watching_tv IS NOT NULL " \
"GROUP BY watching_tv ORDER BY watching_tv").df()

computer_frequency = duckdb.sql("SELECT computer_internet, COUNT(*) " \
"AS frequency, ROUND(COUNT(*) * 100.0 / SUM(COUNT(*)) OVER (), 2) " \
"AS percentage FROM hrs WHERE computer_internet IS NOT NULL " \
"GROUP BY computer_internet ORDER BY computer_internet").df()

exercise_frequency = duckdb.sql("SELECT exercising, COUNT(*) " \
"AS frequency, ROUND(COUNT(*) * 100.0 / SUM(COUNT(*)) OVER (), 2) " \
"AS percentage FROM hrs WHERE exercising IS NOT NULL " \
"GROUP BY exercising ORDER BY exercising").df()

walking_frequency = duckdb.sql("SELECT walking, COUNT(*) " \
"AS frequency, ROUND(COUNT(*) * 100.0 / SUM(COUNT(*)) OVER (), 2) " \
"AS percentage FROM hrs WHERE walking IS NOT NULL " \
"GROUP BY walking ORDER BY walking").df()

education_frequency = duckdb.sql("SELECT education_training, COUNT(*) " \
"AS frequency, ROUND(COUNT(*) * 100.0 / SUM(COUNT(*)) OVER (), 2) " \
"AS percentage FROM hrs WHERE education_training IS NOT NULL " \
"GROUP BY education_training ORDER BY education_training").df()

falling_asleep_frequency = duckdb.sql("SELECT trouble_falling_asleep, COUNT(*) " \
"AS frequency, ROUND(COUNT(*) * 100.0 / SUM(COUNT(*)) OVER (), 2) " \
"AS percentage FROM hrs WHERE trouble_falling_asleep IS NOT NULL " \
"GROUP BY trouble_falling_asleep ORDER BY trouble_falling_asleep").df()

waking_night_frequency = duckdb.sql("SELECT waking_at_night, COUNT(*) " \
"AS frequency, ROUND(COUNT(*) * 100.0 / SUM(COUNT(*)) OVER (), 2) " \
"AS percentage FROM hrs WHERE waking_at_night IS NOT NULL " \
"GROUP BY waking_at_night ORDER BY waking_at_night").df()

waking_early_frequency = duckdb.sql("SELECT waking_too_early, COUNT(*) " \
"AS frequency, ROUND(COUNT(*) * 100.0 / SUM(COUNT(*)) OVER (), 2) " \
"AS percentage FROM hrs WHERE waking_too_early IS NOT NULL " \
"GROUP BY waking_too_early ORDER BY waking_too_early").df()

rested_frequency = duckdb.sql("SELECT feeling_rested, COUNT(*) " \
"AS frequency, ROUND(COUNT(*) * 100.0 / SUM(COUNT(*)) OVER (), 2) " \
"AS percentage FROM hrs WHERE feeling_rested IS NOT NULL " \
"GROUP BY feeling_rested ORDER BY feeling_rested").df()

display(reading_frequency)
display(tv_frequency)
display(computer_frequency)
display(exercise_frequency)
display(walking_frequency)
display(education_frequency)
display(falling_asleep_frequency)
display(waking_night_frequency)
display(waking_early_frequency)
display(rested_frequency)
```

### HRS Summary by Age Group

Next, we will create `grouped_hrs`, which summarizes the HRS data by age group. This is not the same as ATUS which means when we compare data we will have to divide even furthur, however HRS focuses on older population which means we dont have data for less than 25yr old.

For each age group, we calculate the number of participants and summarize the main leisure, sleep, memory, and sedentary-behavior variables. This allows us to identify general differences in leisure activities and health outcomes across age groups. It is important to highlight averages of HRS response codes, not average minutes or percentages.

- 25 - 44
- 45 - 64
- 65+

```python
grouped_hrs = duckdb.sql("SELECT age_group, COUNT(*) " \
"AS num_participants, ROUND(AVG(reading), 2) " \
"AS avg_reading, ROUND(AVG(watching_tv), 2) " \
"AS avg_tv, ROUND(AVG(computer_internet), 2) " \
"AS avg_computer, ROUND(AVG(exercising), 2) " \
"AS avg_exercising, ROUND(AVG(walking), 2) " \
"AS avg_walking, ROUND(AVG(trouble_falling_asleep), 2) " \
"AS avg_trouble_falling_asleep, ROUND(AVG(feeling_rested), 2) " \
"AS avg_feeling_rested, ROUND(AVG(immediate_recall), 2) " \
"AS avg_immediate_recall, ROUND(AVG(delayed_recall), 2) " \
"AS avg_delayed_recall, ROUND(AVG(sitting_hours), 2) " \
"AS avg_sitting_hours FROM (SELECT *, CASE WHEN age BETWEEN 15 AND 24 THEN '15-24' WHEN age BETWEEN 25 AND 44 THEN '25-44' " \
"WHEN age BETWEEN 45 AND 64 THEN '45-64' ELSE '65+' END " \
"AS age_group FROM hrs WHERE age IS NOT NULL) " \
"GROUP BY age_group ORDER BY CASE age_group WHEN '15-24' THEN 1 WHEN '25-44' THEN 2 WHEN '45-64' THEN 3 ELSE 4 END").df()

display(grouped_hrs)
```

### Creating Total Sitting Time

The HRS records daily sitting time using separate variables for hours and minutes. These variables represent two components of the same reported sitting time. We keep
the original variables in our dataset, but remove special survey response codes (negative time) before using them in calculations.

After cleaning, we combine the hour and minute components into a single variable representing total daily sitting time in minutes. This provides a more interpretable measure of sedentary behavior for our later analysis.

```python
hrs["sitting_hours"] = hrs["sitting_hours"].replace(-8, np.nan)
hrs["sitting_minutes"] = hrs["sitting_minutes"].replace(-8, np.nan)
hrs["sitting_total_minutes"] = hrs["sitting_hours"] * 60 + hrs["sitting_minutes"]

display(duckdb.sql("SELECT ROUND(AVG(sitting_total_minutes), 2) " \
"AS avg_sitting_minutes, MIN(sitting_total_minutes) " \
"AS min_sitting_minutes, MAX(sitting_total_minutes) " \
"AS max_sitting_minutes FROM hrs").df())

display(duckdb.sql("SELECT hhid, pn, sitting_hours, sitting_minutes, sitting_total_minutes FROM hrs WHERE sitting_total_minutes > 1440 ORDER BY sitting_total_minutes DESC").df())
```

After combining the sitting-hours and sitting-minutes variables, we found several values that cannot represent actual daily sitting time. For example, some respondents have values of 98 or 99 hours, and one observation produces more than 24 hours of sitting in a single day.

To prevent these special or invalid values from affecting our summary statistics, we create cleaned versions of the sitting-time variables. Valid sitting hours must fall between 0 and 24, while the minutes component must fall between 0 and 59. We then calculate total sitting time and treat totals greater than 1,440 minutes (24 hours) as missing.

```python
hrs["sitting_hours_clean"] = hrs["sitting_hours"].where(hrs["sitting_hours"].between(0, 24), np.nan)
hrs["sitting_minutes_clean"] = hrs["sitting_minutes"].where(hrs["sitting_minutes"].between(0, 59), np.nan)
hrs["sitting_total_minutes"] = hrs["sitting_hours_clean"] * 60 + hrs["sitting_minutes_clean"]
hrs.loc[hrs["sitting_total_minutes"] > 1440, "sitting_total_minutes"] = np.nan

display(duckdb.sql("SELECT ROUND(AVG(sitting_total_minutes), 2) " \
"AS avg_sitting_minutes, MIN(sitting_total_minutes) " \
"AS min_sitting_minutes, MAX(sitting_total_minutes) " \
"AS max_sitting_minutes FROM hrs").df())

display(duckdb.sql("SELECT COUNT(*) AS invalid_sitting_values FROM hrs WHERE sitting_total_minutes > 1440").df())
```

### Cleaning Sleep Variables

The HRS sleep variables contain three substantive response categories, coded as `1`, `2`, and `3`, as well as additional special response codes such as `-8`, `8`, and `9`. Because these special codes do not represent actual levels of sleep difficulty or restfulness, we convert them to missing values before calculating averages or using the variables in exploratory analysis. We keep the original variables and create cleaned versions for analysis.

```python
hrs["trouble_falling_asleep_clean"] = hrs["trouble_falling_asleep"].where(hrs["trouble_falling_asleep"].isin([1, 2, 3]), np.nan)
hrs["waking_at_night_clean"] = hrs["waking_at_night"].where(hrs["waking_at_night"].isin([1, 2, 3]), np.nan)
hrs["waking_too_early_clean"] = hrs["waking_too_early"].where(hrs["waking_too_early"].isin([1, 2, 3]), np.nan)
hrs["feeling_rested_clean"] = hrs["feeling_rested"].where(hrs["feeling_rested"].isin([1, 2, 3]), np.nan)

display(duckdb.sql("SELECT trouble_falling_asleep_clean, COUNT(*) AS frequency FROM hrs GROUP BY trouble_falling_asleep_clean ORDER BY trouble_falling_asleep_clean").df())
display(duckdb.sql("SELECT waking_at_night_clean, COUNT(*) AS frequency FROM hrs GROUP BY waking_at_night_clean ORDER BY waking_at_night_clean").df())
display(duckdb.sql("SELECT waking_too_early_clean, COUNT(*) AS frequency FROM hrs GROUP BY waking_too_early_clean ORDER BY waking_too_early_clean").df())
display(duckdb.sql("SELECT feeling_rested_clean, COUNT(*) AS frequency FROM hrs GROUP BY feeling_rested_clean ORDER BY feeling_rested_clean").df())
```

### Cleaning Leisure Activity Variables

The HRS leisure variables measure how frequently respondents participate in activities such as reading, television viewing, computer use, exercise, walking, and educational
activities. These variables use response codes from 1 through 7. We inspect and clean these variables so that only valid response codes from 1 through 7 are retained. Any other numeric values would be treated as missing.

```python
hrs["reading_clean"] = hrs["reading"].where(hrs["reading"].between(1, 7), np.nan)
hrs["watching_tv_clean"] = hrs["watching_tv"].where(hrs["watching_tv"].between(1, 7), np.nan)
hrs["computer_internet_clean"] = hrs["computer_internet"].where(hrs["computer_internet"].between(1, 7), np.nan)
hrs["exercising_clean"] = hrs["exercising"].where(hrs["exercising"].between(1, 7), np.nan)
hrs["walking_clean"] = hrs["walking"].where(hrs["walking"].between(1, 7), np.nan)
hrs["education_training_clean"] = hrs["education_training"].where(hrs["education_training"].between(1, 7), np.nan)

display(duckdb.sql("SELECT reading_clean, COUNT(*) AS frequency FROM hrs GROUP BY reading_clean ORDER BY reading_clean").df())
display(duckdb.sql("SELECT watching_tv_clean, COUNT(*) AS frequency FROM hrs GROUP BY watching_tv_clean ORDER BY watching_tv_clean").df())
display(duckdb.sql("SELECT exercising_clean, COUNT(*) AS frequency FROM hrs GROUP BY exercising_clean ORDER BY exercising_clean").df())
```

### Cleaning Cognition Variables

The HRS cognition variables measure the number of words correctly recalled during the immediate and delayed word-recall tasks. Valid scores range from 0 to 10 words. To ensure that unexpected values do not affect our analysis, we create cleaned versions that retain only scores between 0 and 10. Missing responses remain recorded as `NaN`.

```python
hrs["immediate_recall_clean"] = hrs["immediate_recall"].where(hrs["immediate_recall"].between(0, 10), np.nan)
hrs["delayed_recall_clean"] = hrs["delayed_recall"].where(hrs["delayed_recall"].between(0, 10), np.nan)
display(duckdb.sql("SELECT immediate_recall_clean, COUNT(*) AS frequency FROM hrs GROUP BY immediate_recall_clean ORDER BY immediate_recall_clean").df())
display(duckdb.sql("SELECT delayed_recall_clean, COUNT(*) AS frequency FROM hrs GROUP BY delayed_recall_clean ORDER BY delayed_recall_clean").df())
```
